### Encapsulation

Encapsulation with Getter and Setter methods and
Public, protected, private variable

```python
class Person:
    def __init__(self, name, age, gender):
        self.__name = name  #Private variable  : you can't access out of the class
        self._age = age #Protected variable  : you can't access out of the class : u can access for derived class
        self.gender = gender #Public variable


    def get_name(person):
        return person.__name
```

```python
class Employee(Person):
    def __init__(self, name, age, gender):
        super().__init__(name, age, gender)
```

```python
pers = Person("sai", 12, "Male")
print(pers.gender)
emp = Employee("sai", 34, "Male")
print(emp._age)
# print(pers.get_name(pers))
```

```python
##Encapsulation based on the getter and setter methods for Private variables

class Person:
    def __init__(self, name, age, gender):
        self.__name = name  #Private variable  : you can't access out of the class


    def get_name(self):
        return self.__name
    
    def set_name(self, name):
        self.__name = name

p = Person("sai", 22, "Male")

print(p.get_name())
p.set_name("R")
print(p.get_name())
```
