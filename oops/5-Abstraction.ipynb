```python
from abc import ABC, abstractmethod

class Vehicle:
    def drive(self):
        print("Drive method is called!")
    @abstractmethod
    def start_engine(self):
        pass
```

```python
class Car(Vehicle):
    def start_engine(self):
        print("Car engine started!")

    def operate_vehicle(vehicle):
        vehicle.start_engine()
        vehicle.drive()
        
car = Car()
car.operate_vehicle()
```
