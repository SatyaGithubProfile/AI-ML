```python
class Error(Exception):
    pass

class dobException(Error):
    pass
```

```python
year = int(input("Enter you DOB..."))
age = 2024 - year
try:
    if age >= 20 and age <= 30:
        print("Hurray! you are elligible")

    else :
        raise dobException
except dobException:
    print("Oops! your age must be in b/w 20-30..")
```
