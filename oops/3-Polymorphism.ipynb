## Polymorphism

### Method Overriding
Method overriding allows a child to provide a specific implementation of a method that is already defined in its parent class

```python
class Animal:
    def speak(self):
        return "sound of the animal"

class Dog(Animal):
    def speak(self):
        return "Woof!"

class Cat(Animal):
    def speak(self):
        return "Meow!"
    
def anima_speak(animal):
    print(animal.speak())

dog = Dog()
print(dog.speak())
anima_speak(dog)
```
