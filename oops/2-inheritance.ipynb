```python
## Inheritance (Single inheritance)

##parent class

class Car :
    def __init__(self, windows, doors, engineType):
        self.windows = windows
        self.doors = doors
        self.engineType = engineType

    def drive(self):
        print(f"The person will drive the {self.engineType} car.")
```

```python
car1 = Car(4, 5, "petrol")
car1.drive()
```

```python
class Tesla(Car):   # car class is inherited with the Tesla
    def __init__(self, windows, doors, engineType, is_self_drive):
        super().__init__(windows, doors, engineType)   # this is how we can intialize car class elements using super keyword
        self.is_self_drive = is_self_drive
    
    def selfDriving(self):
        print(f"Tesla supports the self driving : {self.is_self_drive}")
```

```python
tesla1 = Tesla(4, 5, "petrol", True)
tesla1.drive()
```

```python
## Multiple Inheritance
class Animal:
    def __init__(self, name):
        self.name = name
    def speak(self):
        print("Subclass must implement this method")
    
class Pet:
    def __init__(self, owner):
        self.owner = owner
    
class Dog(Animal, Pet):
    def __init__(self, name, owner):
        Animal.__init__(self, name)
        Pet.__init__(self, owner)
    def speak(self):
        return f"{self.name} say wooof!"

DogC = Dog("sai", "krish" )
print(DogC.speak())
print(f"Owner : {DogC.owner}")
```
