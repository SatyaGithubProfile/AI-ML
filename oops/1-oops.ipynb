```python
## Modeling a Bank account

class BankAccount:
    def __init__(self, ownerName, balance = 0):
        self.name = ownerName
        self.balance = balance
    
    def deposit(self, amount):
        self.balance += amount
        print(f"{amount} added. Total Balance: {self.balance}")
    
    def withdraw(self, amount):
        if amount > self.balance:
            print("Oops, Insufficient Funds!")
        else:
            self.balance -= amount
            print(f"{amount} deducted. Remaining balance: {self.balance}")
    def getBalance(self):
        print(f"Available balance : {self.balance}")
```

```python

## Make some events to bank 
Account = BankAccount("sai", 1000)
Account.getBalance()
```

```python
Account.withdraw(100)
```

```python
Account.deposit(100)
```
