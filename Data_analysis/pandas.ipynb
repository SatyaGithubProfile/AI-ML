```python
import pandas as pd
```

```python
# Series
data =[1,2,3,4,5]
series = pd.Series(data)
print("Series \n",series)
```

```python
## Ceate a series from dictionary

data = {'a':1, 'b':2, 'c':3}
series_dict = pd.Series(data)
print(series_dict)
```

```python
data = [1,2,3,4,5]
index = ['a','b','c','d','e']
print(pd.Series(data, index= index))
```

```python
## Dataframe

## Create a dataframe from a dictionary of list
data = {
    'Name' : ['Sai', 'John', 'Jack'],
    'Age' : [25, 30, 35],
    'City' : ['Bangalore', 'New York', 'Florida']
}

df = pd.DataFrame(data)
print(df)
```

```python
import numpy as np
print(np.array(data))
```
