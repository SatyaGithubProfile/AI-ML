## Copy the function    

```python
def welcome():
    print("welcome function called!")
```

```python
# welcome()
storeNewVar = welcome   ## you can store that function to another variable too.
print(storeNewVar())
del welcome  ## to delete the function 
print(storeNewVar())
```

## CLOUSURE
 A closure means a nested function : function inside a function

```python
def main_welcome():
    msg = "Welcome to closure functionality"
    def sub_welcome():
        print(msg)
        print("This fn is inside another fn")
    return sub_welcome()
```

```python
main_welcome()
```

### Decorators

```python
def main_welcome(func):
    msg = "Welcome to closure functionality"
    def sub_welcome():
        func()
        print(msg)
        print("This fn is inside another fn")
    return sub_welcome()
```

```python
def cource_intro():
    print("it's advance")

cource_intro()
```

```python
@main_welcome
def cource_intro():
    print("it's advance")
```

```python
cource_intro
```

```python
def my_dec(func):
    func()
    print('my decorator is called!')
```

```python
@my_dec
def say_hello():
    print("hello")
```

```python
say_hello()
```
