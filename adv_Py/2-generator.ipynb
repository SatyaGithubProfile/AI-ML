## Yield keyword
If you return a funcion with "yield" keyword, then it'll call it as generator

```python
def square(n):
    for i in range(n):
        yield i **2  ## it will store temporarily and return the value
```

```python
square(4)
```

```python
for i in square(4):
    print(i)
```

```python
def multiple(n):
    for i in range(n):
        yield n*i
```

```python
multiple(3)
```

### Large file reading with yield

```python
def readLargeFile(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line
    
```

```python
file_path = 'large_file.txt'
for line in readLargeFile(file_path):
    print(line.strip())
```
