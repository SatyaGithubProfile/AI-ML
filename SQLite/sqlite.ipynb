```python
import sqlite3
```

```python
connection = sqlite3.connect('example.db')
connection
```

```python
cursor = connection.cursor()
```

```python
cursor.execute('''
    Create Table If Not Exists employees(
               id Integer Primary key,
               name Text Not Null,
               age Integer,
               department text
               )

''')

connection.commit() ## committ changes to execute
```

```python
## Insert the data

cursor.execute('''
    Insert Into employees(name, age, department)
               Values('Bob', 25, 'Engineering')
''')
cursor.execute('''
    Insert Into employees(name, age, department)
               Values('jack', 23, 'J-Engineering')
''')

cursor.execute('''
    Insert Into employees(name, age, department)
               Values('Jhon', 26, 'K-Engineering')
''')
connection.commit()
```

```python
## Query the data from the table
cursor.execute('Select * from employees')
rows = cursor.fetchall()

## print the rows
for r in rows:
    print(r)
```

```python
## Update the data in the db
cursor.execute('''
update employees set age=34 where name = 'Bob'
               ''')

connection.commit()
```

```python
## Delete the data from table 
cursor.execute('''
delete from employees where name = 'jack'

''')
connection.cursor()
```
